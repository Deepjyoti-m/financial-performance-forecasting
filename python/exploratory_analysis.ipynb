```python
import pandas as pd
import matplotlib.pyplot as plt
import os

base_dir = os.path.abspath(os.path.join(os.getcwd(), ".."))
file_path = os.path.join(base_dir, "data", "processed", "financial_cleaned.csv")

df = pd.read_csv(file_path)
df["year_month"] = df["year_month"].astype(str)

print("", df.shape)
df.head()
```

```python
monthly_revenue = df.groupby("year_month")["revenue"].sum().reset_index()

plt.figure(figsize=(14, 5))
plt.plot(monthly_revenue["year_month"], monthly_revenue["revenue"], marker="o", color="steelblue", linewidth=2)
plt.xticks(rotation=45, ha="right")
plt.title("Monthly Revenue Trend", fontsize=14)
plt.xlabel("Month")
plt.ylabel("Revenue ($)")
plt.tight_layout()
plt.savefig("../reports/monthly_revenue_trend.png")
plt.show()
```

```python
product_revenue = df.groupby("product")["revenue"].sum().reset_index().sort_values("revenue", ascending=True)

plt.figure(figsize=(8, 5))
plt.barh(product_revenue["product"], product_revenue["revenue"], color="steelblue")
plt.title("Revenue by Product Category", fontsize=14)
plt.xlabel("Total Revenue ($)")
plt.tight_layout()
plt.savefig("../reports/revenue_by_product.png")
plt.show()
print("")
```

```python
region_profit = df.groupby("region")["profit"].sum().reset_index().sort_values("profit", ascending=True)

plt.figure(figsize=(8, 5))
plt.barh(region_profit["region"], region_profit["profit"], color="seagreen")
plt.title("Profit by Region", fontsize=14)
plt.xlabel("Total Profit ($)")
plt.tight_layout()
plt.savefig("../reports/profit_by_region.png")
plt.show()
print("")
```

```python
margin = df.groupby("product").agg(
    revenue=("revenue", "sum"),
    profit=("profit", "sum")
).reset_index()

margin["gross_margin_pct"] = (margin["profit"] / margin["revenue"] * 100).round(2)

plt.figure(figsize=(8, 5))
plt.bar(margin["product"], margin["gross_margin_pct"], color="coral")
plt.title("Gross Margin % by Product", fontsize=14)
plt.xlabel("Product")
plt.ylabel("Gross Margin (%)")
plt.tight_layout()
plt.savefig("../reports/gross_margin_by_product.png")
plt.show()
print("")
```

```python
df["month"] = df["year_month"].str[5:7].astype(int)

seasonality = df.groupby("month")["revenue"].mean().reset_index()
month_names = ["Jan","Feb","Mar","Apr","May","Jun","Jul","Aug","Sep","Oct","Nov","Dec"]
seasonality["month_name"] = seasonality["month"].apply(lambda x: month_names[x-1])

plt.figure(figsize=(10, 5))
plt.bar(seasonality["month_name"], seasonality["revenue"], color="mediumpurple")
plt.title("Average Revenue by Month (Seasonality)", fontsize=14)
plt.xlabel("Month")
plt.ylabel("Avg Revenue ($)")
plt.tight_layout()
plt.savefig("../reports/seasonality.png")
plt.show()
print("")
```
